## 🔧 Imports

## 📌 Section A — Manual RLGlue Interaction

This section shows how RLGlue calls your agent and environment under the hood.

**Key ideas:**

- `rl_start()` internally calls:
  - `env_start()` to get the first state
  - `agent_start(state)` to choose the first action
- `rl_step()` internally calls:
  - `env_step(action)` to get reward and next state
  - `agent_step(reward, new_state)` (or `agent_end` if terminal)
- You will see the raw values returned at each step.


```python
# RL-Glue warm-up: manual interaction with the environment

# 1. Choose environment and agent classes
EnvClass = ten_arm_env.Environment
AgentClass = main_agent.Agent

# 2. Create RLGlue object
rl_glue = RLGlue(EnvClass, AgentClass)

# 3. Initialize agent and environment
#    - agent_init_info is passed to Agent.agent_init
#    - env_init_info is passed to Environment.env_init
agent_init_info = {"num_actions": 10}  # 10-armed bandit
env_init_info   = {}                    # no extra info needed for this env

rl_glue.rl_init(agent_init_info, env_init_info)

# 4. Start interaction (calls env_start and agent_start internally)
state, action = rl_glue.rl_start()
print(f"Initial state from env_start: {state}")
print(f"Initial action from agent_start: {action}")

# 5. Take several steps (calls env_step + agent_step internally)
print("\nStepping through a few interactions:")
for t in range(5):
    reward, state, action, terminal = rl_glue.rl_step()
    print(f"t={t:2d} | reward={reward:6.3f} | state={state} | action={action} | terminal={terminal}")
```

## 📌 Section B — Running a Full Episode

`rl_episode(max_steps)` automates the interaction for one episode:

- Calls `rl_start()` once
- Repeatedly calls `rl_step()` until:
  - the environment terminates (terminal state), or
  - the maximum number of steps is reached
- Tracks:
  - total reward in the episode (the **return**)
  - number of steps
  - number of episodes so far

You will use exactly this kind of loop in Assignment 1.


```python
# RL-Glue warm-up: running one full episode

# Reinitialize everything so we start fresh
rl_glue.rl_init(agent_init_info, env_init_info)

# Run one episode (up to 1000 steps)
is_terminal = rl_glue.rl_episode(max_steps_this_episode=1000)

print(f"Episode finished. Terminal? {is_terminal}")
print(f"Total reward this episode (return): {rl_glue.rl_return():.3f}")
print(f"Number of steps in this episode: {rl_glue.rl_num_steps()}")
print(f"Episodes so far: {rl_glue.rl_num_episodes()}")
```

```python
# Run many episodes and collect returns

rl_glue.rl_init(agent_init_info, env_init_info)

returns = []
for ep in range(50):
    rl_glue.rl_episode(1000)
    returns.append(rl_glue.rl_return())

print("First 10 returns:", returns[:10])
print("Average return:", sum(returns)/len(returns))
```
